**Importing libraries**

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, GlobalAveragePooling2D
from tensorflow.keras.applications import ResNet50, VGG16, MobileNet
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
```

**Importing files** (Preprocessed data, basic CNN model, transfer learning function)

```python
import data_preprocess as data
import basicCNNmodel as bCNNm
import transfer_learning as TL
```

**Models dictionary**

models to be used: basic CNN model, ResNet, VGG16Net, MobileNet

```python
models = {
    "Sequential CNN": bCNNm.create_sequential_cnn(),
    "ResNet50": TL.create_transfer_model(ResNet50(weights='imagenet', include_top=False, input_shape=(150, 150, 3))),
    "VGG16": TL.create_transfer_model(VGG16(weights='imagenet', include_top=False, input_shape=(150, 150, 3))),
    "MobileNet": TL.create_transfer_model(MobileNet(weights='imagenet', include_top=False, input_shape=(150, 150, 3)))
}
```

```python
IMG_SIZE = (150, 150)
BATCH_SIZE = 32
EPOCHS = 10
```

```python
history_dict = {}
for name, model in models.items():
    print(f"Training {name}...")
    history = model.fit(
        data.train_generator,
        epochs=EPOCHS,
        validation_data=data.test_generator,
        callbacks=[EarlyStopping(patience=3, restore_best_weights=True)]
    )
    history_dict[name] = history
    print(f"✅ {name} training complete!\n")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))

for name, history in history_dict.items():
    plt.plot(history.history['val_accuracy'], label=name)

plt.title('Model Performance Comparison')
plt.xlabel('Epochs')
plt.ylabel('Validation Accuracy')
plt.legend()
plt.show()
```
